```python
from pydantic import BaseModel,Field,EmailStr,ValidationError
class StudentSchema(BaseModel):
    name:str=Field(min_length=3,max_length=100)
    age:int=Field(gt=18,le=60)
    email:EmailStr
    courses:str
try:
    student=StudentSchema(
        name="hi",
        age=18,
        email="kasthu@gmail.com",
        courses="Python")
    #Convert data into json
    print(student.model_dump())
except ValidationError as e:
    print(e)
```

```python
import sqlite3

# Database connection
conn = sqlite3.connect("company.db")

# Cursor object
cursor = conn.cursor()

# Create department table
cursor.execute("""
CREATE TABLE IF NOT EXISTS departments (
    dept_id INTEGER PRIMARY KEY,
    dept_name TEXT,
    location TEXT
)
""")

# Create employee table
cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
    emp_id INTEGER PRIMARY KEY,
    emp_name TEXT,
    salary REAL,
    dept_id INTEGER,
    FOREIGN KEY(dept_id) REFERENCES departments(dept_id)
)
""")

conn.commit()

print("Tables created successfully")
```

```python
# Insert departments
departments = [
    (1, "Python", "Bangalore"),
    (2, "Data Science", "Chennai"),
    (3, "HR", "Hyderabad"),
    (4, "Testing", "Mumbai")
]

cursor.executemany("""
INSERT OR IGNORE INTO departments
VALUES (?, ?, ?)
""", departments)


# Insert employees
employees = [
    (101, "Arun", 50000, 1),
    (102, "Priya", 60000, 2),
    (103, "Kumar", 45000, 1),
    (104, "Divya", 70000, 3),
    (105, "Rahul", 55000, 2)
]

cursor.executemany("""
INSERT OR IGNORE INTO employees
VALUES (?, ?, ?, ?)
""", employees)

conn.commit()

print("Records inserted successfully")
```

```python
query = """
SELECT
    employees.emp_name,
    employees.salary,
    departments.dept_name,
    departments.location
FROM employees
INNER JOIN departments
ON employees.dept_id = departments.dept_id
"""

cursor.execute(query)

records = cursor.fetchall()
print("""
('Rahul', 55000.0, 'Data Science', 'Chennai')


Real-time example:

An HR application needs to display employee details along with their department names.
INNER JOIN retrieves employees who have a matching department.""")
for row in records:
    print(row)
```

```python
query = """
SELECT
    departments.dept_name,
    employees.emp_name
FROM departments
LEFT JOIN employees
ON departments.dept_id = employees.dept_id
"""

cursor.execute(query)

for row in cursor.fetchall():
    print(row)
#An HR manager wants to display all departments, including departments that currently have no employees.
```

```python
query = """
SELECT
    employees.emp_name,
    departments.dept_name
FROM employees
RIGHT JOIN departments
ON employees.dept_id = departments.dept_id
"""

cursor.execute(query)
print("Display all departments, even if no employees are assigned to them.")
for row in cursor.fetchall():
    print(row)
```

```python
query = """
SELECT
    employees.emp_name,
    departments.dept_name
FROM employees
FULL OUTER JOIN departments
ON employees.dept_id = departments.dept_id
"""

cursor.execute(query)

for row in cursor.fetchall():
    print(row)
```

5. SQL Parameterisation using SQLite3
What is parameterisation?
Parameterisation means passing values separately from the SQL query instead of directly inserting user input into the SQL statement.

Why do we use it?
1. Prevents SQL injection attacks.
2. Makes queries safer.
3. Handles user input properly.
4. Improves code readability.
5. Makes dynamic queries easier to write.

```python
#Without parameterisation (unsafe):
emp_id = input("Enter employee ID: ")

query = f"SELECT * FROM employees WHERE emp_id = {emp_id}"

cursor.execute(query)
#This is unsafe because directly inserting user input can allow SQL injection.
```

```python
#With parameterisation (safe):
emp_id = input("Enter employee ID: ")
query = """
SELECT * FROM employees
WHERE emp_id = ?
"""
cursor.execute(query, (emp_id,))
result = cursor.fetchall()
for row in result:
    print(row)
```
