```python
import pymysql

connection = pymysql.connect(
    host="127.0.0.1",
    user="root",
    password="1234",
    port=3306
)

print("Connected to MySQL Server Successfully!")
```

```python
## Check the number of databases available on the server
cursor=connection.cursor()
cursor.execute('show databases')
for i in cursor:
    print(i)
```

```python
import pymysql

connection = pymysql.connect(
    host="127.0.0.1",
    user="root",
    password="1234",
    port=3306
)

cursor = connection.cursor()

cursor.execute("CREATE DATABASE If Not Exists company")

print("Database created successfully")

cursor.close()
connection.close()
```

```python
import pymysql

connection = pymysql.connect(
    host="127.0.0.1",
    user="root",
    password="1234",
    database="company",
    port=3306
)

cursor = connection.cursor()

print("Connected to Company Database")
```

```python
cursor.execute("""
CREATE TABLE emp_1 (
    id INT PRIMARY KEY,
    name VARCHAR(50),
    salary FLOAT
)
""")

cursor.execute("""
INSERT INTO emp_1 VALUES
(1, 'John', 25000),
(2, 'David', 30000),
(3, 'Smith', 35000)
""")

connection.commit()

print("Records inserted successfully")
```

```python
cursor.execute("SELECT * FROM emp_1")

records = cursor.fetchall()

for row in records:
    print(row)

cursor.close()
connection.close()
```
